# Comprehensive Exploratory Data Analysis of Human Trafficking Reports to the Disque 100 Hotline

## Introduction

Disque 100 is Brazil's official human rights hotline operated by the Ministry of Human Rights and Citizenship (MDHC). The service receives, records, analyzes, and forwards reports of alleged human rights violations, including cases related to human trafficking, one of the most severe forms of human exploitation.

This notebook presents a comprehensive exploratory data analysis (EDA) of human trafficking reports registered through the Disque 100 Hotline and made available through the Ministry of Human Rights and Citizenship's open data initiative [1]. The objective is to understand the structure and quality of the dataset, identify temporal, geographic, and sociodemographic patterns, explore relationships among relevant variables, and generate insights that may support public policy, research, and prevention strategies.

In addition to descriptive analyses, this notebook includes data quality assessment, data cleaning procedures, feature engineering, and multivariate exploration to provide a broader understanding of the reported cases and the information captured by the reporting system.

Disque 100 records reports of alleged human rights violations. Therefore, the term _"human trafficking reports"_ is used throughout this notebook to refer to reported or suspected cases rather than confirmed occurrences of human trafficking.

## Objectives

### General Objective

To perform a comprehensive exploratory analysis of human trafficking reports registered through the Disque 100 Hotline, identifying patterns, trends, and characteristics associated with reported cases.

### Specific Objectives

- Assess the quality, completeness, and consistency of the dataset.
- Identify temporal patterns and trends in reporting behavior.
- Analyze the geographic distribution of reports across Brazil.
- Characterize the profiles of victims, suspects, and reporting parties.
- Explore relationships among case characteristics and reported forms of exploitation.
- Create derived variables that may improve analytical understanding of the phenomenon.
- Generate hypotheses that can guide future confirmatory, statistical, and machine learning studies.

## Research Questions

1. What temporal patterns can be observed in human trafficking reports?
2. How are reports distributed across regions, states, and municipalities?
3. What demographic characteristics are most commonly associated with victims and suspects?
4. Are there recurring patterns among reported cases?
5. Which variables appear to be associated with specific forms of exploitation?
6. What data quality issues are present, and how might they affect analytical results?
7. Can feature engineering reveal additional insights not directly observable in the original variables?
8. What hypotheses emerge from exploratory analysis that warrant further investigation?
